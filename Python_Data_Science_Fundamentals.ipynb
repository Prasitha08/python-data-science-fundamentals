## 1. Variables and Data Types
Variables store values. Python automatically identifies the data type.

## 2. Operators

```python
a = 18
b = 4
print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Remainder:", a % b)
```

## 3. Conditional Statements

```python
mark = 78
if mark >= 90:
    grade = "A"
elif mark >= 75:
    grade = "B"
elif mark >= 50:
    grade = "C"
else:
    grade = "Needs improvement"
print("Grade:", grade)
```

## 4. Loops

```python
for day in range(1, 5):
    print("Day", day)

count = 1
while count <= 3:
    print("Count:", count)
    count += 1
```

## 5. Functions

```python
def calculate_average(values):
    if len(values) == 0:
        return 0
    return sum(values) / len(values)

print(calculate_average([80, 75, 90, 85]))
```

## 6. Lists and Dictionaries

```python
subjects = ["Python", "Statistics", "Data Science"]
subjects.append("Machine Learning")
print(subjects)

student = {"name": "Anu", "course": "Data Science", "completed_days": 4}
print(student["name"])
for key, value in student.items():
    print(key, ":", value)
```

## 8. Introduction to Data Science with Pandas
Upload `student_scores.csv` to the same Colab session or keep it in the same folder as this notebook.

```python
import pandas as pd

data = pd.read_csv("student_scores.csv")
display(data.head())
print("Rows and columns:", data.shape)
display(data.describe())
```

## 9. Mini Project: Student Performance Analysis

```python
subjects = ["python", "statistics", "data_science"]
data["average_score"] = data[subjects].mean(axis=1).round(2)
data["result"] = data["average_score"].apply(
    lambda score: "Pass" if score >= 50 else "Needs Improvement"
)
display(data[["name", "average_score", "result"]])
```

## 10. Visualisation with Matplotlib

```python
import matplotlib.pyplot as plt

subject_averages = data[subjects].mean().sort_values(ascending=False)
subject_averages.plot(kind="bar", figsize=(8, 5))
plt.title("Average Score by Subject")
plt.xlabel("Subject")
plt.ylabel("Average Score")
plt.ylim(0, 100)
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

## 11. Findings and Conclusion

```python
print("Average score by subject:")
display(subject_averages.round(2))
print("Highest average:", subject_averages.idxmax())
print("Result counts:")
display(data["result"].value_counts())
```
